```python
from pprint import pprint

def find_next_empty(puzzle):
    # Search the grid for the first empty cell (represented by -1)
    # Returns a tuple (row, col), or (None, None) if the board is complete
    for r in range(9):
        for c in range(9):
            if puzzle[r][c] == -1:
                return r, c
    return None, None  # No empty positions remain

def is_valid(puzzle, guess, row, col):
    # Checks whether placing 'guess' at position (row, col) is valid
    # Ensures the number doesn't already exist in the row, column, or 3x3 subgrid

    # Check row
    if guess in puzzle[row]:
        return False

    # Check column
    if guess in [puzzle[i][col] for i in range(9)]:
        return False

    # Check 3x3 square
    row_start = (row // 3) * 3
    col_start = (col // 3) * 3
    for r in range(row_start, row_start + 3):
        for c in range(col_start, col_start + 3):
            if puzzle[r][c] == guess:
                return False

    return True

def solve_sudoku(puzzle):
    # Solves the Sudoku puzzle using recursive backtracking
    # Modifies the input grid in place and returns True if a solution is found

    row, col = find_next_empty(puzzle)

    # If no empty cells remain, the puzzle is complete
    if row is None:
        return True

    # Attempt to place numbers 1–9 into the empty cell
    for guess in range(1, 10):
        if is_valid(puzzle, guess, row, col):
            puzzle[row][col] = guess  # Make a tentative assignment

            if solve_sudoku(puzzle):
                return True  # Success!

        # Undo the guess (backtrack)
        puzzle[row][col] = -1

    # No valid guess found, puzzle is unsolvable in current configuration
    return False
```

```python
def display_sudoku(grid):
    # Nicely formats the Sudoku grid with 3x3 boundaries
    for i, row in enumerate(grid):
        if i % 3 == 0:
            print("+-------+-------+-------+")
        print("| {} {} {} | {} {} {} | {} {} {} |".format(
            *(str(num) if num != -1 else '.' for num in row)
        ))
    print("+-------+-------+-------+")
```

```python
example_board = [
    [-1, -1, -1,   2, 6, -1,   7, -1, 1],
    [6, 8, -1,   -1, 7, -1,   -1, 9, -1],
    [1, 9, -1,   -1, -1, 4,   5, -1, -1],

    [8, 2, -1,   1, -1, -1,   -1, 4, -1],
    [-1, -1, 4,   6, -1, 2,   9, -1, -1],
    [-1, 5, -1,   -1, -1, 3,   -1, 2, 8],

    [-1, -1, 9,   3, -1, -1,   -1, 7, 4],
    [-1, 4, -1,   -1, 5, -1,   -1, 3, 6],
    [7, -1, 3,   -1, 1, 8,   -1, -1, -1]
]

# Display, solve, and re-display the puzzle
display_sudoku(example_board)
print("Solving...")
if solve_sudoku(example_board):
    print("Solved")
display_sudoku(example_board) 
```
